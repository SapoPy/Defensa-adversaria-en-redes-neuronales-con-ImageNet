```python

# Diccionario para iterar fácilmente
top1_accuracies = {
    "Noisy": (accuracies_Noisy, accuracies_Noisy_50),
    "FGSM": (accuracies_FGSM, accuracies_FGSM_50),
    "RFGSM": (accuracies_RFGSM, accuracies_RFGSM_50),
    "PGD": (accuracies_PDG, accuracies_PDG_50)
}

clean_acc = {
    "34": acc_clean,
    "50": acc_clean_50
}

fig, axes = plt.subplots(2, 2, figsize=(14,10))
axes = axes.flatten()

for i, (attack_name, (acc34, acc50)) in enumerate(top1_accuracies.items()):
    ax = axes[i]
    ax.plot(epsilons, acc34, 'o-', label='Target 34')
    ax.plot(epsilons, acc50, 'o--', label='Target 50')
    ax.axhline(clean_acc["34"], color='gray', linestyle=':', label='Clean 34')
    ax.axhline(clean_acc["50"], color='black', linestyle=':', label='Clean 50')
    
    ax.set_title(f"{attack_name} Top-1 Accuracy")
    ax.set_xlabel("Epsilon")
    ax.set_ylabel("Accuracy")
    ax.set_ylim(0, 1)
    ax.grid(True)
    ax.legend(fontsize=8)

plt.suptitle("Impacto de ataques adversariales sobre Top-1 Accuracy", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```
