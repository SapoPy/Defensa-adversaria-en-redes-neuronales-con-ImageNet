```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
epsilons = [0.05, 0.1, 0.15]
accuracies_FGSM = [0.7866666666666666, 0.68, 0.6533333333333333]
accuracies_RFGSM = [0.84, 0.6466666666666666, 0.3933333333333333]
accuracies_PDG = [0.9066666666666666, 0.8333333333333334, 0.7866666666666666]   
accuracies_Noisy = [0.8866666666666667, 0.84, 0.8266666666666667]
```

```python
acc_clean = 0.9066666666666666
plt.plot(epsilons, accuracies_FGSM, marker='o', label='FGSM')
plt.plot(epsilons, accuracies_RFGSM, marker='o', label='RFGSM')
plt.plot(epsilons, accuracies_PDG, marker='o', label='PDG')
plt.plot(epsilons, accuracies_Noisy, marker='o', label='Gaussian Noise')

# línea de referencia: accuracy sin ataque
plt.plot(epsilons, [acc_clean]*len(epsilons), linestyle='--', color='black', label='Accuracy sin ataque')

plt.title('Epsilon vs Accuracy')
plt.xlabel('Epsilon')
plt.ylabel('Accuracy')
plt.legend()
plt.savefig("graficos/Epsilon vs Accuracy.png", dpi=300, transparent=True, bbox_inches="tight")
plt.show()
```
